# Dataset and basic statistics


## Libraries

```python
from tabulate import tabulate
import pandas as pd
```

## Wheatseed dataset


```python
df = pd.read_csv(
    "/content/wheat-seeds (1).csv",
    header=None,
    names=[
        "Area",
        "Perimeter",
        "Compactness",
        "Length of Kernel",
        "Width of Kernel",
        "Asymmetry Coefficient",
        "Length of Kernel Groove",
        "Class",
    ],
)
```

```python
def summary_stats(
    df: pd.DataFrame,
    mode: bool = False,
    columns: list[str] | None = None,
    exclude_columns: list[str] | None = None
) -> None:

    result = {}
    for data in df.columns:
        if columns and data not in columns:
            continue
        if exclude_columns and data in exclude_columns:
            continue

        data_max = round(df[data].max(), 2)
        data_min = round(df[data].min(), 2)

        result[data] = {
            "class": data,
            "max": data_max,
            "min": data_min,
            "range": round(data_max - data_min, 2),
            "var": round(df[data].var(), 2),
            "std": round(df[data].std(), 2),
            "median": round(df[data].median(), 2),
            "mean": round(df[data].mean(), 2),
        }
        if mode:
            result[data]["mode"] = df[data].mode().values

    print(
        tabulate(
            pd.DataFrame.from_dict(result.values()),
            headers="keys",
            tablefmt="grid",
            showindex=False,
        ),
    )
```

```python
summary_stats(df, mode=True, exclude_columns=['Class'])
```

```python
groupby = df.groupby("Class")
groupby_count = groupby.size().reset_index(name="count")
print(tabulate(groupby_count, headers="keys", tablefmt="grid"))
```

```python
for name, group_data in groupby:
    print(f"\nClass: {name}")
    summary_stats(group_data)
```
